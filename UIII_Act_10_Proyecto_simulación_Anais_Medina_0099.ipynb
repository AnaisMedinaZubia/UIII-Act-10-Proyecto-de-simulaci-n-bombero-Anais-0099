```python
# --- SISTEMA DE IA: GESTIÓN DEL ESCUADRÓN "LOSBOMBEROS" ---

# Inicialización de la estructura de datos: lista vacía
losbomberos = []

# Ciclo infinito para el menú interactivo
while True:
    print("\n========================================")
    print("   CENTRO DE MANDO INTELIGENTE (IA)   ")
    print("========================================")
    print("1. Registrar nuevo bombero")
    print("2. Ver lista del escuadrón")
    print("3. IA: Evaluar aptitud para misión crítica")
    print("4. Salir")

    opcion = input("\nSeleccione una opción (1-4): ")

    # OPCIÓN 1: Registro de datos
    if opcion == "1":
        print("\n[REGISTRO] Ingrese los datos:")
        nombre = input("Nombre del bombero: ")
        edad = int(input("Edad: "))
        años_exp = int(input("Años de experiencia: "))
        rendimiento = int(input("Nivel de rendimiento físico (1-100): "))

        # Se crea un diccionario vacío y se llena dinámicamente [4, 5]
        unbombero = {}
        unbombero["nombre"] = nombre
        unbombero["edad"] = edad
        unbombero["experiencia"] = años_exp
        unbombero["rendimiento"] = rendimiento

        # Se agrega el diccionario a la lista [1]
        losbomberos.append(unbombero)
        print(f"¡{nombre} ha sido integrado con éxito!")

    # OPCIÓN 2: Listar el escuadrón usando ciclo FOR [2, 6]
    elif opcion == "2":
        print("\n--- LISTADO DE PERSONAL ---")
        if len(losbomberos) == 0:
            print("El escuadrón está vacío.")
        else:
            # Encabezado con anchos fijos alineado a la izquierda (<)
            print(f"{'Nombre':<20} | {'Edad':<8} | {'Experiencia':<15} | {'Rendimiento':<12}")
            print("-" * 65) # Línea divisoria perfecta

            # Ciclo for para recorrer la lista de diccionarios
            for b in losbomberos:
                # Damos formato de texto a cada dato para poder aplicar la alineación fija
                nom_str = b['nombre']
                edad_str = f"{b['edad']} años"
                exp_str = f"{b['experiencia']} años"
                rend_str = f"{b['rendimiento']}%"

                # Imprimimos la fila con los mismos anchos del encabezado
                print(f"{nom_str:<20} | {edad_str:<8} | {exp_str:<15} | {rend_str:<12}")

    # OPCIÓN 3: Inteligencia Artificial (Motor de reglas) [7, 8]
    elif opcion == "3":
        if len(losbomberos) == 0:
            print("\n[IA] Error: No hay personal registrado para analizar.")
        else:
            print("\n[IA] Iniciando análisis de aptitud operativa...")
            # La IA recorre cada diccionario de la lista
            for b in losbomberos:
                print(f"\nAnalizando perfil de: {b['nombre']}...")

                # Reglas de decisión de la IA con operadores lógicos y relacionales
                if b['rendimiento'] >= 90 and b['experiencia'] > 5:
                    estatus = "APTITUD: CRÍTICA (Líder de primera línea)."
                elif b['rendimiento'] >= 70 or b['experiencia'] >= 3:
                    estatus = "APTITUD: ALTA (Personal operativo de apoyo)."
                elif b['edad'] > 60:
                    estatus = "APTITUD: RESERVA (Personal de logística/coordinación)."
                else:
                    estatus = "APTITUD: EN ENTRENAMIENTO (No apto para misiones de riesgo)."

                print(estatus)

    # OPCIÓN 4: Salida del sistema
    elif opcion == "4":
        print("\nApagando sistema de IA... ¡Manténganse a salvo, bomberos!")
        print("programa realizado por Anais Medina NL = 40")
        break # Rompe el ciclo while True para cerrar el programa [3]

    # Manejo de opciones no válidas
    else:
        print("\n[Error] Opción no válida. Intente de nuevo.")

print("\n--- Sistema fuera de línea ---")
```
