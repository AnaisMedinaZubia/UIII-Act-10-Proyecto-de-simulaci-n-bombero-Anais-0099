#### Lista bombero

```python
# --- SISTEMA DE IA: GESTIÓN DE ESCUADRÓN LOSBOMBEROS ---

# Lista lineal plana
losbomberos = []

# Ciclo infinito para mantener el menú interactivo
while True:
    print("\n========================================")
    print("   CENTRO DE MANDO INTELIGENTE (IA)   ")
    print("========================================")
    print("1. Registrar nuevo bombero")
    print("2. Consultar lista de escuadrón")
    print("3. IA: Analizar aptitud para misiones críticas")
    print("4. Salir del sistema")

    opcion = input("\nSeleccione una opción (1-4): ")

    if opcion == "1":
        print("\n[REGISTRO] Ingrese los datos del bombero:")
        nombre = input("Nombre: ")
        experiencia = int(input("Años de experiencia: "))
        misiones = int(input("Misiones cumplidas: "))
        salud = int(input("Nivel de salud física (1-100): "))

        # Se agregan los datos de forma lineal
        losbomberos.append(nombre)
        losbomberos.append(experiencia)
        losbomberos.append(misiones)
        losbomberos.append(salud)
        print(f"¡{nombre} ha sido integrado al escuadrón!")

    elif opcion == "2":
        print("\n--- LISTADO ACTUAL DEL ESCUADRÓN ---")
        if len(losbomberos) == 0:
            print("No hay bomberos registrados.")
        else:
            # Encabezado alineado: Nombre (20 caracteres), Experiencia (15) y Salud (10)
            print(f"{'Bombero':<20} | {'Experiencia':<15} | {'Salud':<10}")
            print("-" * 55) # Línea divisoria

            # Recorremos la lista de 4 en 4
            for i in range(0, len(losbomberos), 4):
                nombre_b = losbomberos[i]
                exp_b = f"{losbomberos[i+1]} años"
                salud_b = f"{losbomberos[i+3]}%"

                # El símbolo :<20 le dice a Python que rellene con espacios hasta tener 20 caracteres de ancho
                print(f"{nombre_b:<20} | {exp_b:<15} | {salud_b:<10}")

    elif opcion == "3":
        print("\n[IA] Iniciando análisis de red neuronal para misiones críticas...")
        if len(losbomberos) == 0:
            print("Error: Sin personal para analizar.")
        else:
            for i in range(0, len(losbomberos), 4):
                b_nombre = losbomberos[i]
                b_experiencia = losbomberos[i+1]
                b_misiones = losbomberos[i+2]
                b_salud = losbomberos[i+3]

                print(f"\nAnalizando a: {b_nombre}...")

                if b_salud >= 90 and b_experiencia > 5:
                    diagnostico = "ESTADO: LÍDER DE MISIÓN. Apto para rescate extremo."
                elif b_salud >= 70 or b_misiones > 10:
                    diagnostico = "ESTADO: OPERATIVO. Apto para apoyo en primera línea."
                else:
                    diagnostico = "ESTADO: RESERVA. Se recomienda entrenamiento adicional."

                print(diagnostico)

    elif opcion == "4":
        print("\nApagando sistema de IA... ¡Manténganse a salvo, bomberos!")
        print("Programa realizado por Anais Medina NL = 40")
        break

    else:
        print("\n[Error] Opción no válida. Intente de nuevo.")

print("\n--- Sistema fuera de línea ---")
```
